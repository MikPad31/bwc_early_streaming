## Helper functions

```python
def summarise_col(df, col):
    counts = df[col].value_counts(dropna=False)
    pct = df[col].value_counts(dropna=False, normalize=True).mul(100)
    summary = pd.DataFrame({
        "count": counts,
        "pct": pct.round(2)
    })
    print(summary)
```

```python
def natural_sort_key(s):
    parts = re.split(r"(\d+)", str(s))
    return [int(p) if p.isdigit() else p for p in parts]
```

## Target distribution

What the values mean:

- 0: Fail BWC
- 1: Pass BWC, streamed to Fighter
- 2: Pass BWC, streamed to Transport
- 3: Pass BWC, streamed to Heli

The binary target is fairly balanced: 57.06% pass vs 42.94% fail. That means pass/fail modelling is not heavily affected by class imbalance, though a simple majority-class baseline would still get about 57% accuracy.

The multiclass target is more uneven. Fail is the largest class (42.94%), followed by Fighter (28.67%), Rotary (19.40%), and Transport (8.99%). The small Transport class will probably be the hardest to model well, so multiclass evaluation should pay more attention to macro F1 and per-class recall, not just overall accuracy.

At the batch level, there is quite a lot of variation in both pass rates and stream mix. Some batches have much higher pass rates than others, and the proportions of Fighter, Rotary, and Transport also shift across cohorts. This points to possible cohort effects and distribution shift, which could help explain why earlier models did not generalise well to unseen data.

Some of these swings may be due to small batch sizes, so not every difference should be overinterpreted. Still, the overall batch-to-batch variation is large enough to justify batch-aware validation. A random row-wise train/test split would likely give an overly optimistic result, so future-batch or batch-based validation is a better choice.

## Feature distribution

```python
def _existing_lesson_feature_cols(
    df: pd.DataFrame,
    lessons: Iterable[str],
    feature_suffixes: Iterable[str] = FEATURE_SUFFIXES,
) -> dict[str, dict[str, str]]:
    """
    Return mapping:
        {
            "GH 1": {
                "score": "GH 1_score",
                "missing": "GH 1_missing",
                ...
            },
            ...
        }
    Only includes columns that actually exist in df.
    """
    lesson_cols: dict[str, dict[str, str]] = {}

    for lesson in lessons:
        feature_map: dict[str, str] = {}
        for suffix in feature_suffixes:
            col = f"{lesson}_{suffix}"
            if col in df.columns:
                feature_map[suffix] = col

        if feature_map:
            lesson_cols[lesson] = feature_map

    return lesson_cols


def _safe_numeric(series: pd.Series) -> pd.Series:
    """Convert to numeric where possible."""
    return pd.to_numeric(series, errors="coerce")


def _safe_bool_rate(series: pd.Series) -> float:
    """
    Return proportion True among all non-missing values.
    If all values are missing, return pd.NA.
    """
    s = series.dropna()
    if s.empty:
        return pd.NA

    # Works for pandas BooleanDtype / bool / object booleans
    return float((s == True).mean())  # noqa: E712


def _safe_value_rate(series: pd.Series, value: object) -> float:
    """
    Return proportion equal to `value` among all non-missing values.
    If all values are missing, return pd.NA.
    """
    s = series.dropna()
    if s.empty:
        return pd.NA
    return float((s == value).mean())


def summarize_key_features_by_col(
    merged: pd.DataFrame,
    lessons_to_keep: list[str],
    batch_col: str = "Batch",
) -> pd.DataFrame:
    """
    Create one long summary table with one row per Batch x Lesson.
    """
    if batch_col not in merged.columns:
        raise KeyError(f"'{batch_col}' not found in dataframe.")

    lesson_feature_map = _existing_lesson_feature_cols(merged, lessons_to_keep)

    if not lesson_feature_map:
        raise ValueError("No lesson feature columns found for the supplied lessons.")

    summary_rows: list[dict[str, object]] = []

    for batch_value, batch_df in merged.groupby(batch_col, dropna=False):
        n_students = len(batch_df)

        for lesson, cols in lesson_feature_map.items():
            row: dict[str, object] = {
                f"{batch_col}": batch_value,
                "lesson": lesson,
                "n_students": n_students,
            }

            # ---- score summary ----
            if "score" in cols:
                s = _safe_numeric(batch_df[cols["score"]])
                row["score_non_null_n"] = int(s.notna().sum())
                row["score_missing_n"] = int(s.isna().sum())
                row["score_mean"] = s.mean()
                row["score_std"] = s.std()
                row["score_median"] = s.median()
                row["score_min"] = s.min()
                row["score_max"] = s.max()

            # ---- adjusted score summary ----
            if "adjscore" in cols:
                s = _safe_numeric(batch_df[cols["adjscore"]])
                row["adjscore_non_null_n"] = int(s.notna().sum())
                row["adjscore_mean"] = s.mean()
                row["adjscore_std"] = s.std()
                row["adjscore_median"] = s.median()
                row["adjscore_min"] = s.min()
                row["adjscore_max"] = s.max()

            # ---- missingness ----
            if "missing" in cols:
                s = batch_df[cols["missing"]]
                row["missing_rate"] = _safe_bool_rate(s)


            # ---- flags ----
            if "failed" in cols:
                row["fail_rate"] = _safe_bool_rate(batch_df[cols["failed"]])

            summary_rows.append(row)

    summary_df = pd.DataFrame(summary_rows).sort_values(
        [f"{batch_col}", "lesson"],
        kind="stable",
    )

    return summary_df.reset_index(drop=True)
```
